```python
import duckdb
con = duckdb.connect("../data/football.duckdb", read_only=True)
```

```python
con.sql("""
    DESCRIBE raw.players
""").show()
```

```python
con.sql("""
    SELECT * FROM raw.players
    LIMIT 10
""").show()
```

```python
con.sql("""
    DESCRIBE raw.competitions
""").show()
```

```python
con.sql("""
SELECT competition_id, country_name, name
FROM raw.competitions
ORDER BY country_name
""").show(max_rows=100)
```

```python
con.sql("""
SELECT competition_id, country_name, name
FROM raw.competitions
WHERE LOWER(country_name) = 'poland'
""").show(max_rows=100)
```

```python
con.sql("""
DESCRIBE raw.games
""").show()
```

```python
con.sql("""
SELECT season,
COUNT (*) AS total_games
FROM raw.games
WHERE competition_id = 'PL1'
GROUP BY season
ORDER BY season DESC
""").show(max_rows=100)
```

```python
con.sql("""
SELECT season,
MIN(date) AS first_game_date,
MAX(date) AS last_game_date,
FROM raw.games
WHERE competition_id = 'PL1'
GROUP BY season
ORDER BY season DESC
""").show(max_rows=100)
```

```python
con.sql("""
DESCRIBE raw.appearances
""").show()
```

```python
con.sql("""
SELECT 
COUNT(*) AS appearances,
COUNT(DISTINCT game_id) AS total_games,
COUNT(DISTINCT player_id) AS total_players
FROM raw.appearances
WHERE date BETWEEN '2025-07-01' AND '2026-06-30'
AND competition_id = 'NL1'
""").show(max_rows=100)
```

```python
con.sql("""
    SELECT competition_id, COUNT(*) AS appearances
    FROM raw.appearances
    GROUP BY competition_id
    ORDER BY appearances DESC
""").show(max_rows=100)
```
